# **Outliers Analysis**

```python
q1 = df['Weekly_Sales'].quantile(0.25)
q3 = df['Weekly_Sales'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df['Weekly_Sales'] < lower_bound) | (df['Weekly_Sales'] > upper_bound)]
print(outliers)
```

```python
outliers = df[(df['Weekly_Sales'] < lower_bound) | (df['Weekly_Sales'] > upper_bound)]
```

```python
df ['Weekly_Sales_Capped'] = df['Weekly_Sales']. clip(lower=lower_bound,upper=upper_bound)
```

# **Outliers Store-Wise**

```python
def count_outliers(group):
    q1=group.quantile(0.25)
    q3=group.quantile(0.75)
    iqr=q3-q1
    lower_bound=q1-1.5*iqr
    upper_bound=q3+1.5*iqr
    return ((group < lower_bound) | (group > upper_bound)).sum()
outliers_by_store = df.groupby('Store')['Weekly_Sales'].apply(count_outliers)
print(outliers_by_store)
```

## ***Does Unemployment Affect Sales***

```python
Unemployment_corr = df.groupby('Store').apply(lambda x: x['Weekly_Sales'].corr(x['Unemployment']))
print(Unemployment_corr.sort_values())
```

```python
plt.figure(figsize=(15,6))
Unemployment_corr.sort_values().plot(kind='bar')
plt.title('Unemployment Correlation with Weekly Sales')
plt.xlabel('Store')
plt.ylabel('Correlation')
plt.axhline(0)
plt.show()
```
