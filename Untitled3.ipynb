# Feature Engineering for Machine **Learning**

```python
df = df.sort_values(['Store', 'Date']).reset_index(drop=True)
```

```python
df['time_idx'] = df.groupby('Store').cumcount()

df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year

df['WeekOfYear'] = df['Date'].dt.isocalendar().week.astype(int)
```

```python
df['Month_Sin'] = np.sin(
    2 * np.pi * df['Month'] / 12
)

df['Month_Cos'] = np.cos(
    2 * np.pi * df['Month'] / 12
)
```

# **Create Lag Features**

```python
for lag in [1, 2, 4, 8, 12]:
    df[f'Lag_{lag}'] = (
        df.groupby('Store')['Weekly_Sales']
        .shift(lag)
    )
```

```python
df['Rolling_Mean_4'] = (
    df.groupby('Store')['Weekly_Sales']
    .shift(1)
    .rolling(4)
    .mean()
    .reset_index(level=0, drop=True)
)

df['Rolling_Std_4'] = (
    df.groupby('Store')['Weekly_Sales']
    .shift(1)
    .rolling(4)
    .std()
    .reset_index(level=0, drop=True)
)
```

# **Machine Learning Dataset**

```python
features = [
    'Store',
    'time_idx',
    'Month',
    'WeekOfYear',
    'Year',
    'Month_Sin',
    'Month_Cos',
    'Holiday_Flag',
    'Temperature',
    'Fuel_Price',
    'CPI',
    'Unemployment',
    'Lag_1',
    'Lag_2',
    'Lag_4',
    'Lag_8',
    'Lag_12',
    'Rolling_Mean_4',
    'Rolling_Std_4'
]

model_df = df.dropna(subset=features).copy()

X = model_df[features]
y = model_df['Weekly_Sales']
```

# **Time-Based Train/Test Split**

```python
cutoff_date = df['Date'].max() - pd.Timedelta(weeks=11)

train = model_df[model_df['Date'] < cutoff_date]
test = model_df[model_df['Date'] >= cutoff_date]

X_train = train[features]
y_train = train['Weekly_Sales']

X_test = test[features]
y_test = test['Weekly_Sales']

print(X_train.shape)
print(X_test.shape)
```

# **Random Forest Forecasting Model**

```python
model = RandomForestRegressor(
    n_estimators=200,
    max_depth=18,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

# **Model Evaluation**

```python
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)
```

```python
rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_pred)

print("R2 Score:", r2)
```

# **Forecast Next 12 Weeks for Every Store**

```python
history = df[
    [
        'Store',
        'Date',
        'Weekly_Sales',
        'Holiday_Flag',
        'Temperature',
        'Fuel_Price',
        'CPI',
        'Unemployment'
    ]
].copy()

forecast_results = []

for store in sorted(history['Store'].unique()):

    store_history = (
        history[history['Store'] == store]
        .sort_values('Date')
        .copy()
        .reset_index(drop=True)
    )

    for step in range(1, 13):

        next_date = (
            store_history['Date'].max()
            + pd.Timedelta(weeks=1)
        )

        sales_history = (
            store_history['Weekly_Sales'].tolist()
        )

        row = {}

        row['Store'] = store
        row['Date'] = next_date

        row['time_idx'] = len(store_history)

        row['Month'] = next_date.month

        row['WeekOfYear'] = int(
            next_date.isocalendar().week
        )

        row['Year'] = next_date.year

        row['Month_Sin'] = np.sin(
            2 * np.pi * next_date.month / 12
        )

        row['Month_Cos'] = np.cos(
            2 * np.pi * next_date.month / 12
        )

        # Future holiday/external variables are not available
        # in the original data, so use reasonable assumptions.
        row['Holiday_Flag'] = 0

        row['Temperature'] = store_history[
            'Temperature'
        ].iloc[-1]

        row['Fuel_Price'] = store_history[
            'Fuel_Price'
        ].iloc[-1]

        row['CPI'] = store_history[
            'CPI'
        ].iloc[-1]

        row['Unemployment'] = store_history[
            'Unemployment'
        ].iloc[-1]

        row['Lag_1'] = sales_history[-1]
        row['Lag_2'] = sales_history[-2]
        row['Lag_4'] = sales_history[-4]
        row['Lag_8'] = sales_history[-8]
        row['Lag_12'] = sales_history[-12]

        row['Rolling_Mean_4'] = np.mean(
            sales_history[-4:]
        )

        row['Rolling_Std_4'] = np.std(
            sales_history[-4:],
            ddof=1
        )

        input_data = pd.DataFrame([row])

        prediction = model.predict(
            input_data[features]
        )[0]

        row['Weekly_Sales'] = prediction

        forecast_results.append({
            'Store': store,
            'Date': next_date,
            'Forecast_Weekly_Sales': prediction
        })

        store_history = pd.concat(
            [
                store_history,
                pd.DataFrame([{
                    'Store': store,
                    'Date': next_date,
                    'Weekly_Sales': prediction,
                    'Holiday_Flag': 0,
                    'Temperature': row['Temperature'],
                    'Fuel_Price': row['Fuel_Price'],
                    'CPI': row['CPI'],
                    'Unemployment': row['Unemployment']
                }])
            ],
            ignore_index=True
        )
```
